Imports

```python
import joblib 
import pandas as pd
```

upload model and threshold

```python
model=joblib.load("telco_churn_model.pkl")
threshold=joblib.load("telco_churn_threshold.pkl")
```

Predict function

```python
def predict_churn(customer):
    probability=model.predict_proba(customer)[:,1][0]
    prediction=int(probability>=threshold)

    return probability,prediction
```

Test new customer

```python
new_customer = pd.DataFrame({
    "customerID":[1],
    "gender": ["Male"],
    "SeniorCitizen": [0],
    "Partner": ["Yes"],
    "Dependents": ["No"],
    "tenure": [12],
    "PhoneService": ["Yes"],
    "MultipleLines": ["No"],
    "InternetService": ["Fiber optic"],
    "OnlineSecurity": ["No"],
    "OnlineBackup": ["No"],
    "DeviceProtection": ["No"],
    "TechSupport": ["No"],
    "StreamingTV": ["Yes"],
    "StreamingMovies": ["Yes"],
    "Contract": ["Month-to-month"],
    "PaperlessBilling": ["Yes"],
    "PaymentMethod": ["Electronic check"],
    "MonthlyCharges": [85.5],
    "TotalCharges": [1026.0]
})
probability,prediction=predict_churn(new_customer)
print("Probability",probability)
print("Prediction","Churn" if prediction==1 else "No Churn")
```
